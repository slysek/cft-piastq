# Direct PCSS Bell job

This example submits one logical Bell run with **2,000 shots**. `cft-piastq` splits it into ten sequential PCSS parts of 200 shots and returns exact aggregated counts. Provide the PCSS token only at runtime; never save it in this notebook.

## Clone and install

```bash
git clone https://github.com/slysek/cft-piastq.git
cd cft-piastq
python -m pip install -e ".[direct]"
```

````python
import getpass
import os

from qiskit import QuantumCircuit

from cft_piastq import PiastQClient, PiastQSampler
````

````python
token = os.environ.get("PCSS_TOKEN") or getpass.getpass("PCSS token: ")
````

````python
bell = QuantumCircuit(2, 2, name="bell")
bell.h(0)
bell.cx(0, 1)
bell.measure([0, 1], [0, 1])
print(bell.draw("text"))
````

````python
client = PiastQClient(mode="direct", token=token, verbose=False)
sampler = PiastQSampler(
    client.backend,
    options={"cft_job_name": "direct-bell-2000"},
)
````

````python
job = sampler.run(bell, shots=2000)
result = job.result(timeout=1800)
````

````python
counts = job.counts()[0]
print(counts)
assert sum(counts.values()) == 2000
````
